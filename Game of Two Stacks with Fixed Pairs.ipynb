```python
def gameOfTwoStacks(x, a, b):
    
    i = 0  
    j = 0  
    max_pairs = 0
    current_sum = 0
    
    
    while i < len(a) and current_sum + a[i] <= x:
        current_sum += a[i]
        i += 1

   
    max_pairs = i 
    
    while j < len(b):
        current_sum += b[j]
        j += 1
        
        
        while current_sum > x and i > 0:
            i -= 1
            current_sum -= a[i]
        
        
        if current_sum <= x:
            max_pairs = max(max_pairs, i + j)
    
    return max_pairs


x = 10
a = [4, 2, 4, 6, 1]
b = [2, 1, 8, 5]
print(gameOfTwoStacks(x, a, b)) 
```

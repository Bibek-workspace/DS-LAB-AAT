```python
from collections import deque

def splitQueue(queue):
    even_queue = deque()  
    odd_queue = deque()   
    
  
    while queue:
        element = queue.popleft()
        
        if element % 2 == 0:
            even_queue.append(element)
        else:
            odd_queue.append(element)
    
    return even_queue, odd_queue


queue = deque([1, 2, 3, 4, 5, 6, 7, 8, 9])
even_queue, odd_queue = splitQueue(queue)

print("Even Queue:", list(even_queue))  
print("Odd Queue:", list(odd_queue))    
```
